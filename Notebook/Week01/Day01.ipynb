```python
# Cell 1
# Mục đích: đọc đường dẫn thư mục Olist từ .env (không viết cứng vào code vì repo công khai)
import os
from pathlib import Path
import pandas as pd
from dotenv import load_dotenv, find_dotenv

load_dotenv(find_dotenv())                  # tìm .env từ thư mục hiện tại đi lên, nạp vào biến môi trường
OLIST_DIR = Path(os.getenv("OLIST_DIR", ""))  # đọc OLIST_DIR, đổi thành đối tượng Path
print("Thư mục tồn tại:", OLIST_DIR.exists())  # mong đợi: True
```

```python
# Cell 2
# Mục đích: xem 9 file Olist có tên gì, nặng bao nhiêu, có những cột nào (chỉ đọc dòng tiêu đề)
files = sorted(OLIST_DIR.glob("*.csv"))   # liệt kê mọi file .csv, xếp theo tên
print("Số file:", len(files))             # in trước khi lặp: mong đợi 9 (nếu 0 thì vòng for chạy 0 lần, không báo lỗi)

for f in files:
    cols = pd.read_csv(f, nrows=0).columns.tolist() # nrows=0: chỉ đọc dòng tiêu đề, không đọc dữ liệu
    size_mb = f.stat().st_size / 1024 / 1024          # byte -> MB
    print(f"{f.name} | {size_mb:.2f} MB | {len(cols)} cột")
    print("   ", cols)
```
